```python
import requests
import pandas as pd
```

```python
# 1. Armamos los parámetros del pedido a la API
url = "https://archive-api.open-meteo.com/v1/archive"
params = {
    "latitude": -34.6037,
    "longitude": -58.3816,
    "start_date": "2024-01-01",
    "end_date": "2024-01-31",
    "daily": "temperature_2m_min,temperature_2m_max,precipitation_sum,precipitation_hours",
    "timezone": "America/Argentina/Buenos_Aires"  # para que las fechas coincidan con el huso horario local
}
```

```python
# 2. Hacemos el pedido GET a la API
response = requests.get(url, params=params)
```

```python
# 3. Chequeamos que haya salido bien antes de seguir
if response.status_code == 200:
    data = response.json()
    print("Pedido exitoso")
else:
    print(f"Error {response.status_code}: {response.text}")
```

```python
# 4. Convertimos la parte 'daily' del JSON en un DataFrame
df_clima = pd.DataFrame(data["daily"])
```

```python
# 5. Renombramos columnas para que sean más claras al hacer el join después
df_clima = df_clima.rename(columns={
    "time": "fecha",
    "temperature_2m_min": "temp_min",
    "temperature_2m_max": "temp_max",
    "precipitation_sum": "lluvia_mm",
    "precipitation_hours": "horas_lluvia"
})
```

```python
# 6. Convertimos 'fecha' a datetime para que sea consistente con tu dataset de Ecobici
df_clima["fecha"] = pd.to_datetime(df_clima["fecha"])

print(df_clima.head())
print(f"\nTotal de filas: {len(df_clima)}")
```

```python
# Guardar el DataFrame como CSV local
df_clima.to_csv('../data/processed/ecobici_clima_enero_2024.csv', index=False)
print("CSV guardado")
```
