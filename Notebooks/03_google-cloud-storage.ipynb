```python
from google.cloud import storage
```

```python
# 1. Nombre del bucket y del archivo
bucket_name = "ecobici-portfolio-nahir"
ruta_local = "../data/processed/ecobici_clima_enero_2024.csv"
nombre_en_bucket = "clima/ecobici_clima_enero_2024.csv"  # organizado en una "carpeta" clima/ dentro del bucket
```

```python
# 2. Conectamos con GCS (usa las credenciales que ya autenticaste con gcloud)
client = storage.Client(project="ecobici-portfolio")
bucket = client.bucket(bucket_name)
```

```python
# 3. Subimos el archivo
blob = bucket.blob(nombre_en_bucket)
blob.upload_from_filename(ruta_local)

print(f"Archivo subido a gs://{bucket_name}/{nombre_en_bucket}")
```
